## Research Question

Which batting performance metrics have the strongest relationship with a player's overall impact?

```python
impact_corr = df[["runs", "average", "strike_rate", "fours", "sixes", "impact"]].corr()["impact"].sort_values(ascending=False)

impact_corr
```

```python
top_players = df.nlargest(5, "impact")[["batsman", "team", "impact"]]

top_players
```

## Key Insights

- Strike rate has the strongest positive correlation with player impact (0.728), indicating a strong association between scoring efficiency and overall impact in this dataset.
- Sixes have a similarly strong positive correlation with impact (0.723), suggesting that aggressive boundary scoring is strongly associated with higher impact.
- Total runs show a moderate positive correlation with impact (0.453), while fours have a weaker positive correlation (0.309).
- Batting average has almost no linear correlation with impact (-0.064) in this dataset.
- The five highest-impact players are Vaibhav Sooryavanshi, Abhishek Sharma, Ishan Kishan, Sanju Samson, and Priyansh Arya.

```python
plt.figure(figsize=(8, 5))
sns.barplot(data=top_players, x="impact", y="batsman")
plt.title("Top 5 Players by Impact")
plt.xlabel("Impact")
plt.ylabel("Player")
plt.show()
```

```python
impact_corr.drop("impact").sort_values().plot(kind="barh", figsize=(8, 5))
plt.title("Correlation of Batting Metrics with Player Impact")
plt.xlabel("Correlation")
plt.ylabel("Metric")
plt.show()
```

## Conclusion

This analysis explored the relationship between batting performance metrics and overall player impact.

The correlation analysis helped identify which metrics show the strongest association with player impact, while the visualizations provided a clearer understanding of these relationships.

The analysis also highlighted the top-performing players based on the Impact metric.

These findings are specific to the 15-player dataset used in this study and should not be generalized to all IPL players without a larger and more representative dataset.

## Limitations

- The dataset contains only 15 player records.
- Correlation does not imply causation.
- The analysis is based only on the metrics available in the dataset.
- A larger dataset covering more players and seasons could provide more reliable conclusions.
